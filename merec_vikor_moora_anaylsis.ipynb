```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from docx import Document
from docx.shared import Inches
doc = Document()
doc.add_heading('Elektrikli Araç Analiz Raporu', 0)
# 1. Görüntüleme Ayarları (Sayıların normal (decimal) görünmesi için)
pd.options.display.float_format = '{:.6f}'.format

# 2. Dosyayı Oku
df = pd.read_excel("ev.xlsx", sheet_name=0)

# 3. Veri Yapısını Ayrıştır
# m: Satır sayısı (Alternatifler), n: Sütun sayısı (Kriterler)
# Alternatif isimleri 1. sütunda, 2. satırdan başlar
alternatifler = df.iloc[1:, 0].values
kriterler = df.columns[1:]

# Yönler: İlk satırdaki veriler (maks/min)
yonler_str = df.iloc[0, 1:].values

# Karar Matrisi (X): Sayısal veriler
X = df.iloc[1:, 1:].apply(pd.to_numeric, errors='coerce').fillna(0).values
m, n = X.shape

# 4. Görselleştirme İçin Tabloyu Hazırla
# Yönler satırını içeren bir DataFrame oluştur
yon_df = pd.DataFrame([yonler_str], columns=kriterler, index=['YÖNLER'])

# Araç verilerini içeren DataFrame
veri_df = pd.DataFrame(X, columns=kriterler, index=alternatifler)

# İkisini birleştir
karar_matrisi = pd.concat([yon_df, veri_df])

print(f"Başarılı: {m} adet araç ve {n} adet kriter yüklendi.\n")
display(karar_matrisi)
# MEREC'e özgü normalizasyon (nij)
# Formül: 
# Fayda (maks) ise: min(x) / x
# Maliyet (min) ise: x / max(x)
# DOĞRU MEREC NORMALİZASYONU
norm_merec = np.zeros(X.shape)

for j in range(n):
    if yonler_str[j] == 1: # FAYDA (Maks)
        # Formül: x_ij / max(x_j) -> En iyi değer 1 olur.
        norm_merec[:, j] =np.min(X[:, j]) / (X[:, j] + 1e-9)
    else: # MALİYET (Min)
        # Formül: min(x_j) / x_ij -> En düşük (en iyi) değer 1 olur.
        norm_merec[:, j] = X[:, j] / (np.max(X[:, j]) + 1e-9)
# Sonucu tablo olarak görselleştir
norm_merec_df = pd.DataFrame(norm_merec, index=alternatifler, columns=kriterler)

print("--- 2. ADIM: MEREC NORMALİZE MATRİSİ ---")
display(norm_merec_df)
# 1. Her bir alternatif için toplam performans skorunun (Si) hesaplanması
# Formül: Si = ln(1 + (1/n * sum(|ln(nij)|)))
S = np.log(1 + (1/n * np.sum(np.abs(np.log(norm_merec + 1e-9)), axis=1)))

# 2. Her bir kriterin çıkarılma etkisinin (Ej) hesaplanması
E = np.zeros(n)
for j in range(n):
    # j. kriteri matristen geçici olarak çıkar
    norm_X_removed = np.delete(norm_merec, j, axis=1)
    
    # Kalan kriterlerle (n-1) yeni performans skoru (S'j) hesapla
    S_j = np.log(1 + (1/(n) * np.sum(np.abs(np.log(norm_X_removed + 1e-9)), axis=1)))
    
    # Orijinal S ile yeni S_j arasındaki mutlak farkların toplamı
    E[j] = np.sum(np.abs(S_j - S))

# 3. Nihai Ağırlıkların (wj) hesaplanması
weights = E / np.sum(E)

# Sonuçları tablo olarak görselleştir
agirlik_df = pd.DataFrame({
    'Kriter': kriterler,
    'Ağırlık (%)': weights * 100
})

print("--- 3. ADIM: MEREC KRİTER AĞIRLIKLARI ---")
display(agirlik_df.sort_values(by='Ağırlık (%)', ascending=False))
# --- 4. ADIM: MİN-MAX NORMALİZASYONU ---
# Tüm kriterleri Fayda (1=En İyi, 0=En Kötü) eksenine çekiyoruz
norm_minmax = np.zeros(X.shape)
for j in range(n):
    c_min, c_max = np.min(X[:, j]), np.max(X[:, j])
    if yonler_str[j] == 1: # Fayda (Menzil, Sürat, Tork, Batarya)
        norm_minmax[:, j] = (X[:, j] - c_min) / (c_max - c_min)
    else: # Maliyet (Fiyat, Şarj Süreleri, Enerji Tüketimi)
        norm_minmax[:, j] = (c_max - X[:, j]) / (c_max - c_min)

# --- 5. ADIM: VIKOR HESAPLAMALARI ---
# İdeal noktadan uzaklık matrisi (1 en iyiyse, uzaklık = 1 - değer)
dist_matrix = 1 - norm_minmax

# S_vik: Maksimum Grup Faydası (Ağırlıklı toplam uzaklık)
S_vikor = np.sum(weights * dist_matrix, axis=1)

# R_vik: Minimum Bireysel Pişmanlık (Ağırlıklı maksimum uzaklık)
R_vikor = np.max(weights * dist_matrix, axis=1)

# Q İndeksi Hesabı (v=0.5)
v = 0.5
S_star, S_minus = np.min(S_vikor), np.max(S_vikor)
R_star, R_minus = np.min(R_vikor), np.max(R_vikor)

# Sıfıra bölünme hatasını engellemek için 1e-9 ekliyoruz
Q = v * (S_vikor - S_star) / (S_minus - S_star + 1e-9) + \
    (1 - v) * (R_vikor - R_star) / (R_minus - R_star + 1e-9)

# --- 1. MOORA ORİJİNAL VEKTÖR NORMALİZASYONU ---
# Formül: x_ij / karekök( sütun kareleri toplamı )
# m: alternatif sayısı, n: kriter sayısı
m, n = X.shape
v_norm = np.zeros((m, n))

for j in range(n):
    # Sütunun karelerinin toplamının karekökü
    payda = np.sqrt(np.sum(X[:, j]**2))
    v_norm[:, j] = X[:, j] / (payda + 1e-9)

# --- 2. MOORA SKOR HESABI (Fayda - Maliyet) ---
moora_skor = np.zeros(m)

for i in range(m):
    fayda_toplam = 0
    maliyet_toplam = 0
    for j in range(n):
        if yonler_str[j] == 1: # Maks (Menzil, Tork vb.)
            fayda_toplam += v_norm[i, j] * weights[j]
        else: # Min (Fiyat, Şarj vb.)
            maliyet_toplam += v_norm[i, j] * weights[j]
    
    # MOORA Temel Formülü: Fayda - Maliyet
    moora_skor[i] = fayda_toplam - maliyet_toplam

# --- SONUÇLARI BİRLEŞTİRME VE TABLO ---
nihai_tablo = pd.DataFrame({
    'VIKOR_Q': Q,
    'MOORA_Skor': moora_skor
}, index=alternatifler)

# Sıralamalar (VIKOR'da Küçük Q en iyi, MOORA'da Büyük Skor en iyi)
nihai_tablo['VIKOR_Sıra'] = nihai_tablo['VIKOR_Q'].rank(ascending=True).astype(int)
nihai_tablo['MOORA_Sıra'] = nihai_tablo['MOORA_Skor'].rank(ascending=False).astype(int)

print("--- NİHAİ KARŞILAŞTIRMALI SIRALAMA TABLOSU ---")
display(nihai_tablo.sort_values('VIKOR_Sıra'))
# ==========================================
# 5. GÖRSELLEŞTİRME (PROFESYONEL PAKET)
# ==========================================
plt.figure(figsize=(15, 5))

# Grafik 1: Kriter Ağırlıkları
plt.subplot(1, 2, 1)
sorted_w_idx = np.argsort(weights)
plt.barh(kriter_adlari := kriterler[sorted_w_idx], weights[sorted_w_idx]*100, color='teal')
plt.title('MEREC Kriter Ağırlıkları')
for i, v in enumerate(weights[sorted_w_idx]*100): plt.text(v, i, f' %{v:.2f}')
plt.tight_layout()
plt.savefig("g1.png", dpi=300)
doc.add_heading('1. Kriter Önem Dereceleri (MEREC)', level=1)
doc.add_picture("g1.png", width=Inches(6))
plt.show()

# Grafik 2: Sıralama Karşılaştırma
plt.subplot(1, 2, 2)
for alt in alternatifler:
    plt.plot(['VIKOR', 'MOORA'], [nihai_tablo.loc[alt, 'VIKOR_Sıra'], nihai_tablo.loc[alt, 'MOORA_Sıra']], marker='o', label=alt)
plt.gca().invert_yaxis()
plt.title('Yöntemler Arası Sıralama Değişimi')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.savefig("g2.png", dpi=300)
doc.add_heading('2. Yöntemler Arası Sıralama Karşılaştırması', level=1)
doc.add_picture("g2.png", width=Inches(6))
plt.show()
# 3. KÜMÜLATİF PERFORMANS (Stacked Bar)
# ==========================================
# Her aracın hangi kriterden ne kadar puan topladığını gösterir
agirlik_matrisi = v_norm * weights
agirlik_matrisi_df = pd.DataFrame(agirlik_matrisi, index=alternatifler, columns=kriterler)
agirlik_matrisi_df.plot(kind='bar', stacked=True, figsize=(12, 7), colormap='tab20')
plt.title('Alternatiflerin Kriter Bazlı Ağırlıklı Performans Dağılımı', fontsize=14, fontweight='bold')
plt.ylabel('Toplam MOORA Puanı')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', title="Kriterler")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("g3.png", dpi=300)
doc.add_heading('3. Kümülatif Performans Analizi', level=1)
doc.add_picture("g3.png", width=Inches(6))
plt.show()
# ==========================================
# 4. MOORA PERFORMANS DENGESİ VE SIRALAMA (Güncellenmiş)
# ==========================================
moora_grafik_df = pd.DataFrame(agirlik_matrisi, index=alternatifler, columns=kriterler)

# Maliyet kriterlerini negatife çevir
for j, kriter in enumerate(kriterler):
    if yonler_str[j] != 1: 
        moora_grafik_df[kriter] = moora_grafik_df[kriter] * -1

plt.figure(figsize=(14, 8))
ax = moora_grafik_df.plot(kind='bar', stacked=True, colormap='tab20', ax=plt.gca())

# MOORA Skorlarını (yi) siyah nokta olarak ekle
plt.scatter(range(len(alternatifler)), moora_skor, 
            color='black', zorder=5, label='Nihai MOORA Skoru (yi)', 
            edgecolors='white', s=120)

# --- EKLEME: Her sütunun üzerine sıralama numarasını yazdır ---
# nihai_tablo'daki 'MOORA_Sıra' verisini kullanıyoruz
for i, alt in enumerate(alternatifler):
    sira = nihai_tablo.loc[alt, 'MOORA_Sıra']
    plt.text(i, moora_skor[i] + 0.01, f"{sira}.", 
             ha='center', va='bottom', fontsize=12, fontweight='bold', color='black')

plt.axhline(0, color='black', linewidth=1.2, linestyle='-')
plt.ylabel('Ağırlıklı Performans Puanı')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', title="Kriterler ve Skor")
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.tight_layout()

plt.savefig("g4_siralamali.png", dpi=300)
doc.add_heading('4. Kriter Bazlı Performans Dengesi ve Başarı Sıralaması', level=1)
doc.add_picture("g4_siralamali.png", width=Inches(6))
plt.show()
import os


# Masaüstü yolunu otomatik bulur
masaustu_yolu = os.path.join(os.path.join(os.environ['USERPROFILE']), 'Desktop')
dosya_tam_adi = os.path.join(masaustu_yolu, "Elektrikli Araç Analiz Raporu.docx")

# Dosyayı kaydet
doc.save(dosya_tam_adi)

print(f"--- BAŞARILI ---")
print(f"Dosya şu an Masaüstünde: {dosya_tam_adi}")
```
